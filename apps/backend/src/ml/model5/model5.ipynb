```python
def normalize_state(value):
    value = value.strip().lower()

    state_mapping = {
        "up": "Uttar Pradesh",
        "u.p.": "Uttar Pradesh",
        "uttar pradesh": "Uttar Pradesh",
        "mp": "Madhya Pradesh",
        "m.p.": "Madhya Pradesh",
        "bihar": "Bihar",
        "rajasthan": "Rajasthan",
        "delhi": "Delhi"
    }

    return state_mapping.get(value, value.title())


def normalize_income(value):
    value = str(value).lower().strip()

    value = value.replace("₹", "")
    value = value.replace("rs.", "")
    value = value.replace("rs", "")
    value = value.replace(",", "").strip()

    if "lakh" in value:
        number = float(value.replace("lakh", "").strip())
        return number * 100000

    if "lac" in value:
        number = float(value.replace("lac", "").strip())
        return number * 100000

    return float(value)
```

```python
def get_missing_fields(profile):

    missing_fields = []

    for field, value in profile.items():
        if value is None:
            missing_fields.append(field)

    return missing_fields
```

```python
REQUIRED_FIELDS = [
    "age",
    "occupation",
    "state",
    "annual_income",
    "category",
    "gender"
]


def get_required_missing_fields(profile):

    missing_fields = []

    for field in REQUIRED_FIELDS:
        if profile.get(field) is None:
            missing_fields.append(field)

    return missing_fields
```

```python
FIELD_QUESTIONS = {
    "age": "What is your age?",
    "occupation": "What is your occupation?",
    "state": "Which state do you live in?",
    "annual_income": "What is your annual family income?",
    "category": "Which category do you belong to?",
    "gender": "What is your gender?",
    "education": "What is your highest level of education?",
    "disability": "Do you have any disability?",
    "marital_status": "What is your marital status?"
}


def generate_missing_questions(profile):

    missing_fields = get_required_missing_fields(profile)

    questions = []

    for field in missing_fields:
        questions.append(FIELD_QUESTIONS[field])

    return questions
```

```python
def get_profile_status(profile):

    missing_fields = get_required_missing_fields(profile)

    if len(missing_fields) == 0:
        return {
            "complete": True,
            "missing_fields": [],
            "questions": []
        }

    questions = generate_missing_questions(profile)

    return {
        "complete": False,
        "missing_fields": missing_fields,
        "questions": questions
    }
```
