```python
import re

def tokenize(text):
    """
    Split a transcript into words/numbers while keeping
    Hindi/English words and decimal numbers.
    """
    return re.findall(r'\d+(?:\.\d+)?|[A-Za-zÀ-ÿ]+', str(text))

# Test
test_text = df.loc[1, "user_transcript"]

print(test_text)
print(tokenize(test_text))
```

```python
def normalize(text):
    text = str(text).lower().strip()
    text = re.sub(r'[^\w\s]', ' ', text)
    text = re.sub(r'\s+', ' ', text)
    return text.strip()


print(normalize("Madhya Pradesh."))
print(normalize("Uttar Pradesh"))
print(normalize("Daily Wage Worker"))
```

```python
def find_entity_tokens(tokens, entity_value):
    """
    Find an entity value inside a tokenized sentence.
    Returns the start and end token positions.
    """

    if pd.isna(entity_value):
        return None

    entity_tokens = tokenize(str(entity_value))

    if not entity_tokens:
        return None

    normalized_entity = [normalize(t) for t in entity_tokens]

    for i in range(len(tokens) - len(normalized_entity) + 1):
        window = [normalize(t) for t in tokens[i:i + len(normalized_entity)]]

        if window == normalized_entity:
            return i, i + len(normalized_entity)

    return None
```

```python
def add_entity_labels(tokens, labels, entity_value, entity_type):
    """
    Find an entity in the token list and assign BIO labels.
    """

    position = find_entity_tokens(tokens, entity_value)

    if position is None:
        return

    start, end = position

    # B = beginning of entity
    labels[start] = f"B-{entity_type}"

    # I = inside entity
    for i in range(start + 1, end):
        labels[i] = f"I-{entity_type}"
```

```python
def find_numeric_entity(tokens, value):
    """
    Find a numeric value inside the tokenized transcript.
    """

    if pd.isna(value):
        return None

    # Convert values like 26.0 -> 26
    if isinstance(value, float) and value.is_integer():
        value = int(value)

    value = str(value).strip()

    for i, token in enumerate(tokens):
        # Remove punctuation for comparison
        clean_token = re.sub(r'[^\d.]', '', token)

        if clean_token == value:
            return i, i + 1

    return None
```

```python
# Keywords used to identify disability in the transcript

DISABILITY_KEYWORDS = [
    "divyang",
    "differently abled",
    "differently-abled",
    "disabled",
    "disability"
]


def find_keyword_entity(tokens, keywords):
    """
    Find the first matching keyword/phrase in the token list.
    """
    for keyword in keywords:
        position = find_entity_tokens(tokens, keyword)

        if position is not None:
            return position

    return None
```

```python
def add_keyword_entity(tokens, labels, keywords, entity_type):
    """
    Find and label the first occurrence of a keyword/phrase.
    """
    position = find_keyword_entity(tokens, keywords)

    if position is None:
        return

    start, end = position

    # Don't overwrite an existing entity
    if any(label != "O" for label in labels[start:end]):
        return

    labels[start] = f"B-{entity_type}"

    for i in range(start + 1, end):
        labels[i] = f"I-{entity_type}"
```

```python
def convert_row_to_bio(row):
    """
    Convert one dataset row into token-level BIO annotations.
    """

    text = row["user_transcript"]
    tokens = tokenize(text)

    # Initially every token is outside an entity
    labels = ["O"] * len(tokens)

    # --------------------------------------------------
    # 1. STATE
    # --------------------------------------------------
    add_keyword_entity(
        tokens,
        labels,
        [str(row["state"])],
        "STATE"
    )

    # --------------------------------------------------
    # 2. OCCUPATION
    # --------------------------------------------------
    if not pd.isna(row["occupation"]):
        add_keyword_entity(
            tokens,
            labels,
            [str(row["occupation"])],
            "OCCUPATION"
        )

    # --------------------------------------------------
    # 3. AGE
    # --------------------------------------------------
    age_position = find_numeric_entity(
        tokens,
        row["age"]
    )

    if age_position:
        start, end = age_position

        if all(label == "O" for label in labels[start:end]):
            labels[start] = "B-AGE"

            for i in range(start + 1, end):
                labels[i] = "I-AGE"

    # --------------------------------------------------
    # 4. ANNUAL INCOME
    # --------------------------------------------------
    income_position = find_numeric_entity(
        tokens,
        row["annual_income"]
    )

    if income_position:
        start, end = income_position

        if all(label == "O" for label in labels[start:end]):
            labels[start] = "B-ANNUAL_INCOME"

            for i in range(start + 1, end):
                labels[i] = "I-ANNUAL_INCOME"

    # --------------------------------------------------
    # 5. GENDER
    # --------------------------------------------------
    if not pd.isna(row["gender"]):
        add_keyword_entity(
            tokens,
            labels,
            [str(row["gender"])],
            "GENDER"
        )

    # --------------------------------------------------
    # 6. CASTE CATEGORY
    # --------------------------------------------------
    if not pd.isna(row["caste_category"]):
        add_keyword_entity(
            tokens,
            labels,
            [str(row["caste_category"])],
            "CASTE_CATEGORY"
        )

    # --------------------------------------------------
    # 7. DISABILITY
    # --------------------------------------------------
    if row["is_differently_abled"] == True:
        add_keyword_entity(
            tokens,
            labels,
            DISABILITY_KEYWORDS,
            "DISABILITY"
        )

    return tokens, labels
```

```python
# Check whether each structured entity is actually mentioned
# in its corresponding transcript

entity_match_stats = {
    "AGE": 0,
    "GENDER": 0,
    "OCCUPATION": 0,
    "STATE": 0,
    "ANNUAL_INCOME": 0,
    "CASTE_CATEGORY": 0,
    "DISABILITY": 0
}

for _, row in df.iterrows():

    tokens = tokenize(row["user_transcript"])

    # AGE
    if find_numeric_entity(tokens, row["age"]) is not None:
        entity_match_stats["AGE"] += 1

    # GENDER
    if not pd.isna(row["gender"]):
        if find_entity_tokens(tokens, row["gender"]) is not None:
            entity_match_stats["GENDER"] += 1

    # OCCUPATION
    if find_entity_tokens(tokens, row["occupation"]) is not None:
        entity_match_stats["OCCUPATION"] += 1

    # STATE
    if find_entity_tokens(tokens, row["state"]) is not None:
        entity_match_stats["STATE"] += 1

    # INCOME
    if find_numeric_entity(tokens, row["annual_income"]) is not None:
        entity_match_stats["ANNUAL_INCOME"] += 1

    # CASTE
    if not pd.isna(row["caste_category"]):
        if find_entity_tokens(tokens, row["caste_category"]) is not None:
            entity_match_stats["CASTE_CATEGORY"] += 1

    # DISABILITY
    if row["is_differently_abled"] == True:
        if find_keyword_entity(tokens, DISABILITY_KEYWORDS) is not None:
            entity_match_stats["DISABILITY"] += 1


stats_df = pd.DataFrame(
    list(entity_match_stats.items()),
    columns=["Entity", "Matched_Rows"]
)

stats_df["Match_Percentage"] = (
    stats_df["Matched_Rows"] / len(df) * 100
).round(2)

display(stats_df)
```

```python
# Find rows where the age number occurs more than once
# in the transcript.

ambiguous_age = []

for _, row in df.iterrows():

    tokens = tokenize(row["user_transcript"])

    age = row["age"]

    if isinstance(age, float) and age.is_integer():
        age = int(age)

    age_str = str(age)

    matches = []

    for i, token in enumerate(tokens):
        clean_token = re.sub(r'[^\d.]', '', token)

        if clean_token == age_str:
            matches.append(i)

    if len(matches) > 1:
        ambiguous_age.append({
            "transcript_id": row["transcript_id"],
            "text": row["user_transcript"],
            "age": age,
            "positions": matches
        })

print("Rows with multiple age-number matches:", len(ambiguous_age))

display(pd.DataFrame(ambiguous_age).head(20))
```

```python
def convert_row_to_bio(row):
    """
    Convert one dataset row into token-level BIO annotations.
    """

    text = row["user_transcript"]
    tokens = tokenize(text)

    labels = ["O"] * len(tokens)

    # 1. STATE
    add_keyword_entity(
        tokens,
        labels,
        [str(row["state"])],
        "STATE"
    )

    # 2. OCCUPATION
    if not pd.isna(row["occupation"]):
        add_keyword_entity(
            tokens,
            labels,
            [str(row["occupation"])],
            "OCCUPATION"
        )

    # 3. AGE
    age_position = find_numeric_entity(tokens, row["age"])

    if age_position:
        start, end = age_position

        if all(label == "O" for label in labels[start:end]):
            labels[start] = "B-AGE"

            for i in range(start + 1, end):
                labels[i] = "I-AGE"

    # 4. ANNUAL INCOME
    income_position = find_numeric_entity(tokens, row["annual_income"])

    if income_position:
        start, end = income_position

        if all(label == "O" for label in labels[start:end]):
            labels[start] = "B-ANNUAL_INCOME"

            for i in range(start + 1, end):
                labels[i] = "I-ANNUAL_INCOME"

    # 5. GENDER
    add_keyword_entity(
        tokens,
        labels,
        GENDER_KEYWORDS,
        "GENDER"
    )

    # 6. CASTE CATEGORY
    if not pd.isna(row["caste_category"]):
        add_keyword_entity(
            tokens,
            labels,
            [str(row["caste_category"])],
            "CASTE_CATEGORY"
        )

    # 7. DISABILITY
    if row["is_differently_abled"] == True:
        add_keyword_entity(
            tokens,
            labels,
            DISABILITY_KEYWORDS,
            "DISABILITY"
        )

    return tokens, labels
```

```python
def validate_bio(labels):
    errors = []

    for i, label in enumerate(labels):

        if label.startswith("I-"):
            entity = label[2:]

            if i == 0:
                errors.append((i, label))
                continue

            previous = labels[i - 1]

            if previous not in [f"B-{entity}", f"I-{entity}"]:
                errors.append((i, label))

    return errors


invalid_sequences = []

for item in all_results:

    errors = validate_bio(item["labels"])

    if errors:
        invalid_sequences.append({
            "transcript_id": item["transcript_id"],
            "errors": errors
        })


print("Transcripts with invalid BIO sequences:", len(invalid_sequences))
```
